```python
import os
import openai
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file
openai.api_key = os.environ['OPENAI_API_KEY']
```

```python
import json

def get_stock_price(company):
    """Get the current stock price of a company"""
    stock_info = {
        "company": company,
        "price": "198.45",
        "currency": "USD"
    }
    return json.dumps(stock_info)
```

```python
functions = [
    {
        "name": "get_stock_price",
        "description": "Get the current stock price of a company",
        "parameters": {
            "type": "object",
            "properties": {
                "company": {
                    "type": "string",
                    "description": "The name of the company, e.g. Apple, Tesla, Microsoft"
                }
            },
            "required": ["company"],
        },
    }
]
```

```python
messages=[
    {
        "role":"user",
        "content":"Get the stock price of tesla"
    }
]
```

```python
# Call the ChatCompletion endpoint
response = openai.ChatCompletion.create(
   
    model="gpt-3.5-turbo",
    messages=messages,
    functions=functions
)
```

```python
print(response) 
```

```python
response_message = response["choices"][0]["message"]
```
